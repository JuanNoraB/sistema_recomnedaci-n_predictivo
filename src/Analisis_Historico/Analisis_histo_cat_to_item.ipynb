```python
import ast
import pandas as pd

# Configuración pandas
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)

# ============================================================================
# TRABAJAR CON COPIA PARA NO MODIFICAR ORIGINAL
# ============================================================================
df_valid = df_predicciones.copy()

# ============================================================================
# PROCESAR GAPS Y FILTRAR CV VÁLIDO
# ============================================================================
def parse_gaps(x):
    try:
        if pd.isna(x) or x == '[]' or x == '':
            return []
        x_clean = str(x).strip('[]')
        if x_clean == '':
            return []
        return [float(val.strip()) for val in x_clean.split(',')]
    except:
        return []

df_valid['gaps_list'] = df_valid['Ciclos_gaps_originales_dias'].apply(parse_gaps)
df_valid['gaps_len'] = df_valid['gaps_list'].apply(len)
df_valid['Ciclos_cv'] = df_valid['Ciclos_cv'].apply(lambda x: 1 if x == 999 else x)
df_valid['cv_categoria'] = df_valid['Ciclos_cv'].apply(lambda x: 'alto' if x > 0.5 else 'bajo')

# ============================================================================
# RECOLECTAR CASOS EN LISTA
# ============================================================================
casos_lista = []

# CASO A: AMBOS NO_CICLICO
filtro_ambos_no_ciclico = (
    (df_valid['Ciclos_tipo_ciclo'] == 'no_ciclico') & 
    (df_valid['Debug_ciclos_tipo_ciclo_b'] == 'no_ciclico')
)

# A1: gaps_len largo
temp = df_valid[filtro_ambos_no_ciclico].sort_values('gaps_len', ascending=False).head(3).copy()
temp['tipo_caso'] = 'no_ciclico_ambos_gaps_largo'
casos_lista.append(temp)

# A2: score_final alto
temp = df_valid[filtro_ambos_no_ciclico].sort_values('fnn_prob', ascending=False).head(3).copy()
temp['tipo_caso'] = 'no_ciclico_ambos_score_alto'
casos_lista.append(temp)

# CASO B: NO_CICLICO + CORTO_MEDIO
filtro_b = (df_valid['Ciclos_tipo_ciclo'] == 'no_ciclico') & (df_valid['Debug_ciclos_tipo_ciclo_b'] == 'corto_medio')
if filtro_b.sum() > 0:
    # B1: gaps_len largo
    temp = df_valid[filtro_b].sort_values('gaps_len', ascending=False).head(2).copy()
    temp['tipo_caso'] = 'no_ciclico_corto_medio_gaps_largo'
    casos_lista.append(temp)
    
    # B2: CV alto
    temp = df_valid[filtro_b & (df_valid['cv_categoria'] == 'alto')].head(2).copy()
    temp['tipo_caso'] = 'no_ciclico_corto_medio_cv_alto'
    casos_lista.append(temp)
    
    # B3: CV bajo + gaps_len largo
    temp = df_valid[filtro_b & (df_valid['cv_categoria'] == 'bajo')].sort_values('gaps_len', ascending=False).head(2).copy()
    temp['tipo_caso'] = 'no_ciclico_corto_medio_cv_bajo_gaps_largo'
    casos_lista.append(temp)

# CASO C: NO_CICLICO + MEDIANO
filtro_c = (df_valid['Ciclos_tipo_ciclo'] == 'no_ciclico') & (df_valid['Debug_ciclos_tipo_ciclo_b'] == 'mediano')
if filtro_c.sum() > 0:
    # C1: gaps_len largo
    temp = df_valid[filtro_c].sort_values('gaps_len', ascending=False).head(2).copy()
    temp['tipo_caso'] = 'no_ciclico_mediano_gaps_largo'
    casos_lista.append(temp)
    
    # C2: CV alto
    temp = df_valid[filtro_c & (df_valid['cv_categoria'] == 'alto')].head(2).copy()
    temp['tipo_caso'] = 'no_ciclico_mediano_cv_alto'
    casos_lista.append(temp)
    
    # C3: CV bajo + gaps_len largo
    temp = df_valid[filtro_c & (df_valid['cv_categoria'] == 'bajo')].sort_values('gaps_len', ascending=False).head(2).copy()
    temp['tipo_caso'] = 'no_ciclico_mediano_cv_bajo_gaps_largo'
    casos_lista.append(temp)

# CASO D: NO_CICLICO + LARGO
filtro_d = (df_valid['Ciclos_tipo_ciclo'] == 'no_ciclico') & (df_valid['Debug_ciclos_tipo_ciclo_b'] == 'largo')
if filtro_d.sum() > 0:
    # D1: gaps_len largo
    temp = df_valid[filtro_d].sort_values('gaps_len', ascending=False).head(2).copy()
    temp['tipo_caso'] = 'no_ciclico_largo_gaps_largo'
    casos_lista.append(temp)
    
    # D2: CV alto
    temp = df_valid[filtro_d & (df_valid['cv_categoria'] == 'alto')].head(2).copy()
    temp['tipo_caso'] = 'no_ciclico_largo_cv_alto'
    casos_lista.append(temp)
    
    # D3: CV bajo + gaps_len largo
    temp = df_valid[filtro_d & (df_valid['cv_categoria'] == 'bajo')].sort_values('gaps_len', ascending=False).head(2).copy()
    temp['tipo_caso'] = 'no_ciclico_largo_cv_bajo_gaps_largo'
    casos_lista.append(temp)

# ============================================================================
# CONCATENAR TODOS LOS CASOS
# ============================================================================
df_ejemplos_final = pd.concat(casos_lista, ignore_index=True)

print(f"\n📊 Total casos: {len(df_ejemplos_final)}")
print(f"\nDistribución por tipo_caso:")
print(df_ejemplos_final['tipo_caso'].value_counts())


print(df_ejemplos_final[['tipo_caso', 'CODIGO_FAMILIA', 'COD_SUBCATEGORIA', 
                          'Ciclos_tipo_ciclo', 'Debug_ciclos_tipo_ciclo_b',
                          'gaps_len', 'Ciclos_cv', 'cv_categoria', 'score_final']])
```

```python
def seleccionar_entidad_jerarquica(
    df,
    cluster=None,          # None=random, 1-20=específico
    familia=None,          # None=random, 'Q1','Q2','Q3','Q4'=cuartil, valor específico=familia exacta
    subcategoria=None,     # None=random, 'Q1','Q2','max','min'=según transacciones, valor específico=subcat exacta
    item=None,             # None=random, 'Q1','Q2','max','min'=según transacciones, valor específico=item exacto
    random_state=None,
    verbose=True
):
    """
    Selecciona cluster, familia, subcategoría e item de forma jerárquica y retorna la serie de tiempo filtrada.
    
    Parámetros:
    -----------
    df : DataFrame
        DataFrame histórico con columnas: cluster, CODIGO_FAMILIA, COD_SUBCATEGORIA, COD_ITEM, DIM_FACTURA
    cluster : int o None
        Cluster específico (1-20) o None para aleatorio
    familia : str, int o None
        - None: aleatorio
        - 'Q1','Q2','Q3','Q4': cuartil según transacciones
        - valor específico: código de familia exacto
    subcategoria : str, int o None
        - None: aleatorio
        - 'Q1','Q2': cuartil según transacciones (solo mitades)
        - 'max','min': subcategoría con más/menos transacciones
        - valor específico: código de subcategoría exacto
    item : str, int o None
        - None: aleatorio
        - 'Q1','Q2': cuartil según transacciones (solo mitades)
        - 'max','min': item con más/menos transacciones
        - valor específico: código de item exacto
    random_state : int o None
        Semilla para reproducibilidad
    verbose : bool
        Si True, imprime los parámetros seleccionados
    
    Retorna:
    --------
    DataFrame filtrado con la serie de tiempo del cluster/familia/subcategoria/item seleccionado
    """
    import numpy as np
    
    np.random.seed(random_state)
    
    # Guardar argumentos originales
    args_originales = {
        'cluster': cluster,
        'familia': familia,
        'subcategoria': subcategoria,
        'item': item
    }
    
    # 1. SELECCIONAR CLUSTER
    if cluster is None:
        cluster = np.random.choice(df['cluster'].unique())
    
    df_cluster = df[df['cluster'] == cluster].copy()
    
    # 2. SELECCIONAR FAMILIA
    if familia is None:
        # Random
        familia_sel = np.random.choice(df_cluster['CODIGO_FAMILIA'].unique())
    elif isinstance(familia, str) and familia in ['Q1', 'Q2', 'Q3', 'Q4']:
        # Cuartil según transacciones
        trax_por_familia = df_cluster.groupby('CODIGO_FAMILIA')['DIM_FACTURA'].nunique()
        
        # Q1 = más transacciones, Q4 = menos transacciones
        if familia == 'Q1':
            quantiles = [0.75, 1.0]
        elif familia == 'Q2':
            quantiles = [0.5, 0.75]
        elif familia == 'Q3':
            quantiles = [0.25, 0.5]
        else:  # Q4
            quantiles = [0.0, 0.25]
        
        q_min = trax_por_familia.quantile(quantiles[0])
        q_max = trax_por_familia.quantile(quantiles[1])
        familias_en_cuartil = trax_por_familia[(trax_por_familia >= q_min) & (trax_por_familia <= q_max)].index
        familia_sel = np.random.choice(familias_en_cuartil)
    else:
        # Valor específico
        familia_sel = familia
    
    df_familia = df_cluster[df_cluster['CODIGO_FAMILIA'] == familia_sel].copy()
    
    # 3. SELECCIONAR SUBCATEGORÍA (solo Q1, Q2, max, min)
    if subcategoria is None:
        # Random
        subcat_sel = np.random.choice(df_familia['COD_SUBCATEGORIA'].unique())
    elif isinstance(subcategoria, str) and subcategoria in ['Q1', 'Q2', 'max', 'min']:
        trax_por_subcat = df_familia.groupby('COD_SUBCATEGORIA')['DIM_FACTURA'].nunique()
        
        if subcategoria == 'max':
            subcat_sel = trax_por_subcat.idxmax()
        elif subcategoria == 'min':
            subcat_sel = trax_por_subcat.idxmin()
        else:
            # Solo mitades: Q1 = top 50%, Q2 = bottom 50%
            if subcategoria == 'Q1':
                quantiles = [0.5, 1.0]
            else:  # Q2
                quantiles = [0.0, 0.5]
            
            q_min = trax_por_subcat.quantile(quantiles[0])
            q_max = trax_por_subcat.quantile(quantiles[1])
            subcats_en_cuartil = trax_por_subcat[(trax_por_subcat >= q_min) & (trax_por_subcat <= q_max)].index
            subcat_sel = np.random.choice(subcats_en_cuartil)
    else:
        # Valor específico
        subcat_sel = subcategoria
    
    df_subcat = df_familia[df_familia['COD_SUBCATEGORIA'] == subcat_sel].copy()
    
    # 4. SELECCIONAR ITEM (solo Q1, Q2, max, min)
    if item is None:
        # Random
        item_sel = np.random.choice(df_subcat['COD_ITEM'].unique())
    elif isinstance(item, str) and item in ['Q1', 'Q2', 'max', 'min']:
        trax_por_item = df_subcat.groupby('COD_ITEM')['DIM_FACTURA'].nunique()
        
        if item == 'max':
            item_sel = trax_por_item.idxmax()
        elif item == 'min':
            item_sel = trax_por_item.idxmin()
        else:
            # Solo mitades: Q1 = top 50%, Q2 = bottom 50%
            if item == 'Q1':
                quantiles = [0.5, 1.0]
            else:  # Q2
                quantiles = [0.0, 0.5]
            
            q_min = trax_por_item.quantile(quantiles[0])
            q_max = trax_por_item.quantile(quantiles[1])
            items_en_cuartil = trax_por_item[(trax_por_item >= q_min) & (trax_por_item <= q_max)].index
            item_sel = np.random.choice(items_en_cuartil)
    else:
        # Valor específico
        item_sel = item
    
    # Filtrar DataFrame final
    df_resultado = df[
        (df['cluster'] == cluster) &
        (df['CODIGO_FAMILIA'] == familia_sel) &
        (df['COD_SUBCATEGORIA'] == subcat_sel) &
        (df['COD_ITEM'] == item_sel)
    ].copy()
    
    # Imprimir información
    if verbose:
        # Obtener nombres
        nombre_subcat = df_subcat['NOMBRE_SUBCATEGORIA'].iloc[0] if len(df_subcat) > 0 else 'N/A'
        nombre_item = df_resultado['NOMBRE_ITEM'].iloc[0] if len(df_resultado) > 0 else 'N/A'
        
        # Estadísticas de familia completa (todas las subcategorías)
        trax_familia = df_familia['DIM_FACTURA'].nunique()
        num_subcats_familia = df_familia['COD_SUBCATEGORIA'].nunique()
        
        # Estadísticas de subcategoría
        num_items_subcat = df_subcat['COD_ITEM'].nunique()
        trax_subcat = df_subcat['DIM_FACTURA'].nunique()
        
        # Top 5 items de la subcategoría
        top5_items = df_subcat.groupby(['COD_ITEM', 'NOMBRE_ITEM'])['DIM_FACTURA'].nunique().sort_values(ascending=False).head(5)
        top5_list = [(cod, nombre) for (cod, nombre) in top5_items.index]
        
        # Transacciones del item seleccionado
        trax_item = df_resultado['DIM_FACTURA'].nunique() if len(df_resultado) > 0 else 0
        
        print(f"cluster: {args_originales['cluster']} → {cluster}")
        print(f"familia: {args_originales['familia']} → {familia_sel} ({trax_familia} transacciones totales, {num_subcats_familia} subcategorías)")
        print(f"subcategoria: {args_originales['subcategoria']} → {subcat_sel} ({nombre_subcat})")
        print(f"item: {args_originales['item']} → {item_sel} ({nombre_item})")
        print()
        print(f"Subcategoría: {num_items_subcat} items, {trax_subcat} transacciones")
        print(f"Item seleccionado: {trax_item} transacciones")
        print(f"Top 5 items: {top5_list}")
    
    return df_resultado
```
